```python
import os
from faster_whisper import WhisperModel

# المسار الذي يعمل في جهازك (تأكد من اختيار اسم ملف صحيح من القائمة السابقة)
file_name = "34075991-0799455543-20260402142958.wav"
audio_path = os.path.expanduser(f"~/Downloads/audio/{file_name}")

# تأكد من تحميل الموديل أولاً (استخدمنا turbo لتجنب امتلاء الذاكرة)
model = WhisperModel("turbo", device="cuda", compute_type="int8_float16")

# تنفيذ النسخ
segments, info = model.transcribe(audio_path, beam_size=5)

# تحويل الـ generator إلى قائمة لبدء النسخ فوراً وتخزينه في الذاكرة
segments = list(segments)  #

# الآن يمكنك طباعة النتائج
print(f"تم نسخ الملف بنجاح! اللغة: {info.language}")
for segment in segments:
    print("[%.2fs -> %.2fs] %s" % (segment.start, segment.end, segment.text))
```

```python
import os
from faster_whisper import WhisperModel, BatchedInferencePipeline

# تأكد من استخدام المسار الصحيح للملف الموجود في جهازك
file_name = "34075991-0799455543-20260402142958.wav"
audio_path = os.path.expanduser(f"~/Downloads/audio/{file_name}")

# إعداد الموديل مع مراعاة حجم الذاكرة (استخدام int8_float16)
model = WhisperModel("turbo", device="cuda", compute_type="int8_float16")
batched_model = BatchedInferencePipeline(model=model)

# تنفيذ النسخ على دفعات (Batching) لزيادة السرعة
# تم استبدال "audio.mp3" بالمتغير audio_path
segments, info = batched_model.transcribe(audio_path, batch_size=16)

print(f"تم البدء بمعالجة الملف. اللغة المكتشفة: {info.language}")

for segment in segments:
    print("[%.2fs -> %.2fs] %s" % (segment.start, segment.end, segment.text))
```
