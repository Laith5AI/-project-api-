```python
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine, inspect
from sshtunnel import SSHTunnelForwarder

load_dotenv()

def list_tables(engine, label):
    print(f"\n{'='*60}\n{label}\n{'='*60}")
    insp = inspect(engine)
    for schema in insp.get_schema_names():
        if schema in ("pg_catalog", "information_schema"):
            continue
        tables = insp.get_table_names(schema=schema)
        if not tables:
            continue
        print(f"\n[schema: {schema}]")
        for t in tables:
            print(f"  - {t}")

# --- CDP: direct connection ---
cdp_url = (
    f"postgresql+psycopg2://{os.getenv('CDP_POSTGRES_USER')}:"
    f"{os.getenv('CDP_POSTGRES_PASSWORD')}@{os.getenv('CDP_POSTGRES_HOST')}:"
    f"{os.getenv('CDP_POSTGRES_PORT')}/{os.getenv('CDP_POSTGRES_DB')}"
)
cdp_engine = create_engine(cdp_url)
list_tables(cdp_engine, "CDP DATABASE")
cdp_engine.dispose()

# --- LMS: SSH tunnel ---
with SSHTunnelForwarder(
    (os.getenv("LMS_SSH_HOST"), 22),
    ssh_username=os.getenv("LMS_SSH_USER"),
    ssh_pkey=os.getenv("LMS_SSH_KEY_PATH"),
    remote_bind_address=("localhost", int(os.getenv("LMS_DB_PORT", 5432))),
    local_bind_address=("localhost", 0),
) as tunnel:
    lms_url = (
        f"postgresql+psycopg2://{os.getenv('LMS_DB_USER')}:"
        f"{os.getenv('LMS_DB_PASSWORD')}@127.0.0.1:{tunnel.local_bind_port}/"
        f"{os.getenv('LMS_DB_NAME')}"
    )
    lms_engine = create_engine(lms_url)
    list_tables(lms_engine, "LMS DATABASE (via SSH)")
    lms_engine.dispose()
```

```python
from sqlalchemy import text

def explore_table(engine, table_name):
    with engine.connect() as conn:
        rows = conn.execute(text(f"""
            SELECT column_name, data_type
            FROM information_schema.columns
            WHERE table_schema = 'public' AND table_name = :t
            ORDER BY ordinal_position
        """), {"t": table_name}).fetchall()
    print(f"\n--- {table_name} ---")
    for col, dtype in rows:
        print(f"  {col:40s} {dtype}")

cdp_tables = [
    "questions", "skills", "question_skill",
    "topics", "learning_outcomes", "content_learning_outcome",
    "domains", "subjects",
]

cdp_engine = create_engine(cdp_url)
for t in cdp_tables:
    explore_table(cdp_engine, t)
cdp_engine.dispose()
```

```python
import csv
from collections import defaultdict

with SSHTunnelForwarder(
    (os.getenv("LMS_SSH_HOST"), 22),
    ssh_username=os.getenv("LMS_SSH_USER"),
    ssh_pkey=os.getenv("LMS_SSH_KEY_PATH"),
    remote_bind_address=("localhost", int(os.getenv("LMS_DB_PORT", 5432))),
    local_bind_address=("localhost", 0),
) as tunnel:
    lms_url = (
        f"postgresql+psycopg2://{os.getenv('LMS_DB_USER')}:"
        f"{os.getenv('LMS_DB_PASSWORD')}@127.0.0.1:{tunnel.local_bind_port}/"
        f"{os.getenv('LMS_DB_NAME')}"
    )
    lms_engine = create_engine(lms_url)
    with lms_engine.connect() as conn:

        # All assessments (columns)
        assessments = conn.execute(text("""
            SELECT id, title->>'en' AS title FROM assessments ORDER BY id
        """)).fetchall()
        assessment_ids = [a[0] for a in assessments]
        assessment_labels = {a[0]: f"assessment_{a[0]}" for a in assessments}

        # All users via students table (rows)
        users = conn.execute(text("""
            SELECT DISTINCT s.student_id::text, u.id AS user_id
            FROM student_grade_book s
            JOIN students st ON st.id::text = s.student_id
            JOIN users u ON u.id = st.user_id
            ORDER BY u.id
        """)).fetchall()

        # All grade book entries
        grades = conn.execute(text("""
            SELECT student_id::text, assessment_id, mark, full_mark, finished
            FROM student_grade_book
        """)).fetchall()

    lms_engine.dispose()

# Build lookup: (student_id, assessment_id) → score
score_lookup = {}
for student_id, assessment_id, mark, full_mark, finished in grades:
    if full_mark and full_mark > 0:
        score_lookup[(student_id, assessment_id)] = round(mark / full_mark, 4)
    else:
        score_lookup[(student_id, assessment_id)] = 0.0

# Write CSV
out_path = '/home/laith/user_assessment_matrix.csv'
with open(out_path, 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    # Header
    writer.writerow(['user_id', 'student_id'] + [assessment_labels[a] for a in assessment_ids])
    # Rows
    for student_id, user_id in users:
        row = [user_id, student_id]
        for a_id in assessment_ids:
            row.append(score_lookup.get((student_id, a_id), 'NA'))
        writer.writerow(row)

print(f"Saved {len(users):,} users × {len(assessment_ids)} assessments → {out_path}")
```

```python
"""Append-only CSV telemetry for the lesson diagnostic.

Thread-safe, never raises into the calling code. Set the output path via
the DIAGNOSTIC_TELEMETRY_CSV env var (default: ./diagnostic_telemetry.csv).
"""
import csv
import os
import threading
from datetime import datetime, timezone
from pathlib import Path
from typing import Any

_LOCK = threading.Lock()
_CSV_PATH = Path(os.getenv("DIAGNOSTIC_TELEMETRY_CSV", "diagnostic_telemetry.csv"))

FIELDS = [
    "event_ts",                       # ISO-8601 UTC
    "event_type",                     # "answer" | "skill_finalised" | "terminated"
    "session_id",
    "learner_id",
    "lesson_id",
    "phase",                          # initial | binary_search | no_prereqs
    "skill_chain_length",
    "first_lesson_skill_index",
    "current_skill_index",
    "current_skill_id",
    "lo",
    "hi",
    "last_direction",                 # left | right
    "question_id",
    "option_id",
    "is_correct",                     # 0/1
    "current_skill_questions_asked",
    "current_skill_questions_needed",
    "current_skill_correct_count",
    "total_questions_asked",
    "max_total_questions",
    "skill_passed",                   # only on skill_finalised
    "terminated",                     # 0/1
    "termination_reason",
    "failed_prereq_skill_ids",        # pipe-joined
]


def _ensure_header() -> None:
    if _CSV_PATH.exists() and _CSV_PATH.stat().st_size > 0:
        return
    _CSV_PATH.parent.mkdir(parents=True, exist_ok=True)
    with _CSV_PATH.open("w", newline="", encoding="utf-8") as f:
        csv.writer(f).writerow(FIELDS)


def log_event(**kwargs: Any) -> None:
    """Append one row. Silently drops errors so telemetry can never break a request."""
    try:
        row = {k: "" for k in FIELDS}
        row["event_ts"] = datetime.now(timezone.utc).isoformat()
        for k, v in kwargs.items():
            if k not in row:
                continue
            if isinstance(v, bool):
                row[k] = int(v)
            elif isinstance(v, (list, tuple)):
                row[k] = "|".join(str(x) for x in v)
            elif v is None:
                row[k] = ""
            else:
                row[k] = v
        with _LOCK:
            _ensure_header()
            with _CSV_PATH.open("a", newline="", encoding="utf-8") as f:
                csv.writer(f).writerow([row[k] for k in FIELDS])
    except Exception as exc:  # noqa: BLE001
        print(f"[telemetry] write failed: {exc}")
```

```python
import csv
from collections import defaultdict

# Load assessment_lo_map
alo = defaultdict(list)  # assessment_id → [(source_code, weight)]
with open('/home/laith/assessment_lo_map.csv') as f:
    for row in csv.DictReader(f):
        alo[int(row['assessment_id'])].append(
            (row['outcome_source_code'], float(row['weight']))
        )

# Load user_assessment_matrix
with open('/home/laith/user_assessment_matrix.csv') as f:
    reader = csv.DictReader(f)
    umat_rows = list(reader)
    assessment_cols = [c for c in reader.fieldnames if c.startswith('assessment_')]

all_lo_codes = sorted({code for items in alo.values() for code, _ in items})

# Build user × LO matrix
with open('/home/laith/user_lo_matrix.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['user_id', 'student_id'] + all_lo_codes)

    covered = 0
    for row in umat_rows:
        lo_scores = {}
        for col in assessment_cols:
            score = row[col]
            if score == 'NA':
                continue
            a_id = int(col.replace('assessment_', ''))
            if a_id not in alo:
                continue
            for src_code, weight in alo[a_id]:
                prev = lo_scores.get(src_code, 0.0)
                lo_scores[src_code] = prev + float(score) * weight

        out = [row['user_id'], row['student_id']]
        for code in all_lo_codes:
            out.append(round(lo_scores[code], 4) if code in lo_scores else 'NA')
        writer.writerow(out)
        if any(v != 'NA' for v in out[2:]):
            covered += 1

total_lo = len(all_lo_codes)
total_users = len(umat_rows)
filled = sum(1 for r in umat_rows
             for col in assessment_cols
             if r[col] != 'NA')
print(f"Saved: /home/laith/user_lo_matrix.csv")
print(f"Shape: {total_users} users × {total_lo} LOs")
print(f"Users with at least one LO score: {covered}")
```

```python
import csv
from collections import defaultdict

# ── CONFIG ───────────────────────────────────────────────
PASS_THRESHOLD = 0.5   # fraction of full_mark to count as "passed"
# Exclude sessions where student opened and immediately quit (auto-finished with 0)
EXCLUDE_AUTO_ABANDONED = True
# ─────────────────────────────────────────────────────────

with SSHTunnelForwarder(
    (os.getenv("LMS_SSH_HOST"), 22),
    ssh_username=os.getenv("LMS_SSH_USER"),
    ssh_pkey=os.getenv("LMS_SSH_KEY_PATH"),
    remote_bind_address=("localhost", int(os.getenv("LMS_DB_PORT", 5432))),
    local_bind_address=("localhost", 0),
) as tunnel:
    lms_url = (
        f"postgresql+psycopg2://{os.getenv('LMS_DB_USER')}:"
        f"{os.getenv('LMS_DB_PASSWORD')}@127.0.0.1:{tunnel.local_bind_port}/"
        f"{os.getenv('LMS_DB_NAME')}"
    )
    lms_engine = create_engine(lms_url)
    with lms_engine.connect() as conn:

        # All assessments
        all_assessments = [r[0] for r in conn.execute(text(
            "SELECT id FROM assessments ORDER BY id"
        )).fetchall()]

        # All students who ever appeared in grade book
        all_students = [r[0] for r in conn.execute(text(
            "SELECT DISTINCT student_id FROM student_grade_book ORDER BY student_id"
        )).fetchall()]

        # Sessions — ordered per (student, assessment)
        auto_filter = "AND NOT (is_auto_finished = TRUE AND mark = 0)" if EXCLUDE_AUTO_ABANDONED else ""
        sessions = conn.execute(text(f"""
            WITH sessions AS (
                SELECT
                    student_id,
                    assessment_id,
                    started_at,
                    CASE WHEN full_mark > 0
                         THEN mark::float / full_mark
                         ELSE 0
                    END AS score_pct,
                    ROW_NUMBER() OVER (
                        PARTITION BY student_id, assessment_id
                        ORDER BY COALESCE(started_at, created_at)
                    ) AS session_n
                FROM student_grade_book
                WHERE finished = TRUE
                {auto_filter}
            ),
            first_pass AS (
                SELECT student_id, assessment_id, MIN(session_n) AS sessions_until_pass
                FROM sessions
                WHERE score_pct >= :threshold
                GROUP BY student_id, assessment_id
            ),
            attempted AS (
                SELECT DISTINCT student_id, assessment_id FROM sessions
            )
            SELECT
                a.student_id,
                a.assessment_id,
                CASE
                    WHEN fp.sessions_until_pass IS NOT NULL
                        THEN fp.sessions_until_pass::text
                    ELSE 'NC'
                END AS cell_value
            FROM attempted a
            LEFT JOIN first_pass fp
                ON fp.student_id = a.student_id
               AND fp.assessment_id = a.assessment_id
        """), {"threshold": PASS_THRESHOLD}).fetchall()

        # Also get user_id mapping (student_id → user_id)
        student_to_user = dict(conn.execute(text("""
            SELECT s.id::text AS student_id, s.user_id
            FROM students s
        """)).fetchall())

    lms_engine.dispose()

# Build lookup
cell_lookup = {(r[0], r[1]): r[2] for r in sessions}

# Write matrix
out_path = '/home/laith/user_assessment_attempts_matrix.csv'
header = ['student_id', 'user_id'] + [f'assessment_{a}' for a in all_assessments]

with open(out_path, 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(header)
    for sid in all_students:
        row = [sid, student_to_user.get(str(sid), '')]
        for a_id in all_assessments:
            row.append(cell_lookup.get((sid, a_id), 'NA'))
        writer.writerow(row)

# Sanity check
total_cells = len(all_students) * len(all_assessments)
na = sum(1 for sid in all_students for a in all_assessments
         if cell_lookup.get((sid, a), 'NA') == 'NA')
nc = sum(1 for v in cell_lookup.values() if v == 'NC')
n  = sum(1 for v in cell_lookup.values() if v not in ('NA','NC'))

print(f"Saved: {out_path}")
print(f"Shape: {len(all_students)} students × {len(all_assessments)} assessments")
print(f"  NA  (never attempted):        {na:,} / {total_cells:,} ({100*na/total_cells:.1f}%)")
print(f"  NC  (attempted, never passed):{nc:,}")
print(f"  N   (sessions until pass):    {n:,}")
print(f"\nPass threshold: {PASS_THRESHOLD} | Auto-abandoned excluded: {EXCLUDE_AUTO_ABANDONED}")
```
